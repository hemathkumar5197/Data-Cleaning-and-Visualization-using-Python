```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
```

```python
df = pd.read_csv("mtcars.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.describe().sum()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df["hp"])
plt.title("Boxplot of horsepower")
plt.show()
```

```python
plt.figure(figsize=(7,3))
sns.histplot(df["mpg"], bins=10)
plt.title("Distribution of MPG")
plt.show()
```

```python
plt.figure(figsize=(7,3))
sns.scatterplot(x="hp", y="wt", data=df)
plt.title("Horsepower vs Weight")
plt.show()
```

```python
plt.figure(figsize=(7,3))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x="carb", y="disp", data=df)
plt.title("Average Displacement by carburetors")
plt.show()
```
